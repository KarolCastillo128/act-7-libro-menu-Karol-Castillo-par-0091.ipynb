* menu de problemas
1. calcular promedio C1, C2, C3, C4
2. el área de una circunferencia
3. el área de la figura

```python
# menu general de problemas

import math

def vermenu():
    print("\nMENU PRINCIPAL")
    print("1. Ejemplo 2.2 Promedio de 4 calificaciones")
    print("2. Ejemplo 2.4 Area de una circunferencia")
    print("3. Ejemplo 2.6 Area de figura compuesta")
    print("4. Ejemplo 2.8 Distancia entre dos puntos")
    print("5. Ejemplo 2.10 Metros a pulgadas")
    print("6. Salir")
    opcion = int(input("ingrese una opcion: "))
    return opcion

# ------------------- EJEMPLO 2.2 -------------------
def ej22():
    print("\n--- Ejemplo 2.2 ---")
    C1=float(input("ingrese la calificacion 1: "))
    C2=float(input("ingrese la calificacion 2: "))
    C3=float(input("ingrese la calificacion 3: "))
    C4=float(input("ingrese la calificacion 4: "))
    S=C1+C2+C3+C4
    P=S/4
    print("El promedio es: ", P)

# ------------------- EJEMPLO 2.4 -------------------
def ej24():
    print("\n--- Ejemplo 2.4 ---")
    R=float(input("ingrese el radio: "))
    PI=3.1416
    Area=PI*R*R
    print("El area es: ", Area)

# ------------------- EJEMPLO 2.6 -------------------
def ej26():
    print("\n--- Ejemplo 2.6 ---")
    R=float(input("ingrese R: "))
    H=float(input("ingrese H: "))
    C=math.sqrt(H*H - R*R)
    AT=2*(R*C)/2
    PI=3.1416
    AC=(PI*R*R)/2
    Area=AT+AC
    print("El area total es: ", Area)

# ------------------- EJEMPLO 2.8 -------------------
def ej28():
    print("\n--- Ejemplo 2.8 ---")
    X1=float(input("ingrese X1: "))
    Y1=float(input("ingrese Y1: "))
    X2=float(input("ingrese X2: "))
    Y2=float(input("ingrese Y2: "))
    X=X2-X1
    Y=Y2-Y1
    D=math.sqrt(X*X + Y*Y)
    print("La distancia es: ", D)

# ------------------- EJEMPLO 2.10 -------------------
def ej210():
    print("\n--- Ejemplo 2.10 ---")
    CM=float(input("ingrese los metros: "))
    PG=CM/0.0254
    print("Pulgadas a pedir: ", PG)

# ------------------- PROGRAMA PRINCIPAL -------------------
while True:
    op = vermenu()
    if op == 1:
        ej22()
    elif op == 2:
        ej24()
    elif op == 3:
        ej26()
    elif op == 4:
        ej28()
    elif op == 5:
        ej210()
    elif op == 6:
        print("gracias por usar el programa")
        break
    else:
        print("opcion incorrecta")

print("programa realizado por Karol Castillo NL = 22")
print("utilizando IA de colab de google")
```
